```python
cat_names=[]# Finalizing list practice
while True:
    name = input("Enter the name of cat " + str(len(cat_names)+1) + " (or Enter nothing to stop): ")
    if name == "":
     break
    else:
     cat_names = cat_names + [name]#List concatenation
print("The cat names are: ")
for name in cat_names:
    print(" ", name)
```
