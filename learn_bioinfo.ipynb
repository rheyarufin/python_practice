```python
import random

def Score(motifs):
    k = len(motifs[0])
    score = 0
    
    for j in range(k):
        count = {}
        for motif in motifs:
            base = motif[j]
            count[base] = count.get(base, 0) + 1
        
        score += len(motifs) - max(count.values())
    
    return score


def Profile(motifs):
    k = len(motifs[0])
    t = len(motifs) + 4  # Laplace smoothing
    
    profile = [[1]*k for _ in range(4)]
    index = {'A':0,'C':1,'G':2,'T':3}
    
    for motif in motifs:
        for j in range(k):
            profile[index[motif[j]]][j] += 1
    
    for i in range(4):
        for j in range(k):
            profile[i][j] /= t
    
    return profile


def WeightedRandomChoice(probabilities):
    total = sum(probabilities)
    r = random.uniform(0, total)
    
    cumulative = 0
    for i, p in enumerate(probabilities):
        cumulative += p
        if r <= cumulative:
            return i


def ProfileRandomlyGeneratedKmer(text, k, profile):
    index = {'A':0,'C':1,'G':2,'T':3}
    
    probs = []
    kmers = []
    
    for i in range(len(text) - k + 1):
        kmer = text[i:i+k]
        prob = 1
        
        for j in range(k):
            prob *= profile[index[kmer[j]]][j]
        
        probs.append(prob)
        kmers.append(kmer)
    
    chosen_index = WeightedRandomChoice(probs)
    return kmers[chosen_index]


# 🔴 Direct translation of your pseudocode
def GibbsSample(Dna, k, t, N):
    
    # randomly select initial motifs
    Motifs = []
    for dna in Dna:
        i = random.randint(0, len(dna) - k)
        Motifs.append(dna[i:i+k])
    
    BestMotifs = Motifs[:]
    
    for _ in range(N):
        
        # i ← Random(t)
        i = random.randint(0, t-1)
        
        # remove Motifi
        motifs_excluding_i = Motifs[:i] + Motifs[i+1:]
        
        # build profile without Motifi
        profile = Profile(motifs_excluding_i)
        
        # Motifi ← Profile-randomly generated k-mer
        Motifs[i] = ProfileRandomlyGeneratedKmer(Dna[i], k, profile)
        
        # update best motifs
        if Score(Motifs) < Score(BestMotifs):
            BestMotifs = Motifs[:]
    
    return BestMotifs


dna="ATCCTCTATTACCGAGGACTCCATATGTTTGGCCCAACTAGTGGAACTCAAATTATTAGTGCCACTCCTCTGGACATGGTTCTTGCTGTTCTTTCCTTCAAAGGAACGATGCCAGGGTTAGATGGCAGGTTGTAGGGTCTTATCAGTGGGAGCAAACAATTGGGAGCGGTCATCCTAATATTCCTCCCTTCCTAGACATGTTCGTTGGCCAACCAGCCAAATCCATCTCAACCATATACAACGAAGACCAGGCAGGACAATTCGTCTACCCACTTGCCCGTGAGTTGAAGACTGCAGTACGACAACACTTCTCATCCTCTATTACCGA GGACTCCATATGTTTGGCCCAACTAGTGGAACTCAAATTATTAGTGCCACTCCTCTGGACATGGTTCTTGCTGTTCTTTCCTTCAAAGGAACGATGCCAGGGTTAGATGGCAGGTTGTAGGGTCTTATCAGTGGGAGCAAACAATTGGGAGCGGTCATCCTAATATTCCTCCCTTCCTAGACATGTTCGTTGGCCAACCAGCCAAATCCATCTCAACCATATACAACGAAGACCAGGCAGGACAATTCGTCTACCCACTCACGGGTTATAACTCTGCCCGTGAGTTGAAGACTGCAGTACGACAACACTTCTCATCCTCTATTACCGA GCGGGGCAGAGACTCTTTGCATGATGGTCCACGTTTCTCGACGAGGATGCTTTAACCCACTTACACTAACTCTCTAAGGTGGGGCTTAGATCTTTGAGATATTTGTCCATCGCGTTAAATACGAGTCCACTGTCGCATGTCCGACTTCGAACTACCGACTGTTGGAGCAGATAAATACTTGCGTTGCTTATGCAATCAGTGATCCGTAAGTTCGGCGTAAAATTCTGAAAGTGGCTACTAACTGTGGTGATGGCTTAACCAATTATTGAGCTAGCAACTAGCTGATCGAGCTGCGCGATTTGTCTGCAAGCCCAGGGGCGCCTGTGCG GACTGCATAGTACTTTGCTGGAAACCAAATCTTGTGCTCCCACGCCCTCATGCCATTAGAAGAACTCAATACCCGCCCGGACTCGGCCTAGGATATCTATATGCGTGATTACTCCTAAGACCCAATTGCAGTTGCGCACCGCTTATAACTCGATGTTATTAACGCGACGCACAATTGCGCATTATACTGAACCGGAGATCTCAACTGGATACAATTTAGATCTATTTAGATCCTTCAGTGGTAGTGGCTTTGACAGTTTGTGGCGTCTTGGTGTCAAAAGGCTCTCGAAGATTTGCAAGACCCCCCCGAAAGGTTTTTCAACCGCCTC AGTCCCCTACTCTAGTTAGCTGGGGTATAGTTAGATGTAAGTGATGTAGTTGTCAATGTCGGATACGTCCTTTCCCCCGGTGCATTCTGGGGTTCTGGTAATAAACTCATCTAACTATCCAACGCTGATTTCCCGACCAGACGAAAGGATGTCGCTCCGCGCTGAATCATATCGGGCTTCACAGGGAAACTATATAAGCACCACACACTTCAATCGTATAACTTAATATTTATTCCATTGCGGCTTTCAGGCTCTTCTCTCTAGCGCCTTCTTCAGCGCCAAAACCACTTAAAGTAACTCCCGAAGCCACATCACAAACTAAGCTAAC TTGATGAATAGGCTGCCGAATCTCAGGTTGAGAAGGGGTATATCGATTAAGTGTGGATCGCCAACCCCCTCAATCTTGTTGGATAACCCTCCACTTTCCATAACTCACTCGTCCGCAGATCTCAGCTCTCAAGTTTGGTCGGGCAATTACTCCTACAGACATTCAGCAATTCACGCCTTGTAAACCCCATTGGCGACTAAAAATTTATACTCATGGTAGCGACCGACCGTTCACTAACCACCCCTCGGCCCTGACCACCACCGCTAGCTGTAAATACTTGATTCGGGACCGTGCACACTGCTAAATAGGTGTAATGTCTGCACTGGTA GTGCACTTATTATAAGGTAGTTAGCCCTGGTGACACAGCCCAATCTACTCGCAATCATCCCCTAACCGAGCTCCGGGAGTAATGTACGATGGTAGCTCTTATGTGCAGCTATGCAGTGGTACATGGTACCGGCCATATGGCGTAAACCTCCTAACGAGTGGCGGTTATATAACCTGGATGCAAGTAAATATTTAACGCCGTAATGTTACGCCATTAGATTTTGGCAGCTGTGCGCGCACACCCTTCGTCTCTGCCGGGACTTCTAGTGTATGAAGCCCTCGTGGTAAAATACCCCTAATTGCCGGCGCCGGACTCCGCTTATAGCGGA GTAAAAATAGAGTGCAGGCACATCCAATTCGAGCGAGGATTGGCCGGGTTTGGCAGTTCACTCACATGAAATGATCCACGCCCCTTACAAAAAATTACAAGGCCACTACGTATAACTCCGGTCTGGTTGCACCACCCAATGAGGAGCCACTGTACACGACTCTTTGCTGAGGATACGTGGAGAGAGGTCGAGGGTTTTGGGCGAGGGCTTAGTTGTATTAGCCACTTTCTGGGATCGCTTCATAATATTACTTGTCGGCCATAACGCACGGTAGGGTTACCGTGTTCAGGGGTGATAAGTGTCGATACCTATGCTCCCGCAGCCATCC TCTCTGTTTTATGTTACGCTTCCACTTGTACCAGGGATATTCGGTTCTGAGGCAGCCTCTTCCTCACACTCACGTGAAGGGCGTTGATTGATAACGAGCGAGTCCTGTCTCTATCCGAAAAGGGTTTGGTGTTATTATAACTCTGAGATTGGTGGTTTGATTGCCTTTGACTCCGGGGTTCGAATCATTGCGCAGGTGTTAATTACGACGATGCCTTAACGTCCCGAGGCTCACGTTATACTCGCCTTGTATCAACGCGTCGGTAATTTCGAGATCAACCCCGCGCGTTCGGAGGGCCAATTTGAATGTGTCGTGGATTTCCCTGCTT AGACTCGCTCCCGATAGTCCAAATACGCTCAAAATACGTGCACGGAGTTTTGCACGCACGACGATGAGATTTCTTCTTTATTGTGCATCGAGAAGGCTGAGGCTGAATAGTAAAACACGGAGAATGCCTAGAACGGTGACAGTATACGCTTTCTTTTAATAAGCGCGCATTGGGCAGGGTCTGGGCCGGTGACATCCGAGATGTTTATTTCAAATTCAAAAGGGCGGAGGAGCGGGGTGTAATAATGGCGGACTCAGATTCGTTAGAATGACAACAGCGGATTCTTATTATAACTAGTCTCCATAGGAACTACTGCAAGAACCAAATA TCTCTCGTTGACCTCAATGTCCCGGCATTGCATATGCTGCCACAGGACAAAATTAGTTAATCGCCCAAATTGCATTGACGATTTGGATGGGTTGCTGTTTTGACCGCTGCACTGGAGGGGCGCCAGGACGCCACGGATTAATTGGACTGTTCCGAAGTCATAATATCTCCTTCATATACATGGCATGAGGTCCCCATTTTGTTGTCATGCGGGTTGCACCTACTTATTATAACGTCATCGTTAGCTCACCTTCGCACAGATATCCGATTGCCTAAGGTCCAGAACGATATGAGCAAGATTGTGCCGGCGATTTACCAGCGCGCTATGA GATTGACTAACCGACTAGGCTTACCCCGGCTAGGATTATAACACTTATGGAAACTCATCATCCGCTGGACTATACTTAAAGCCCTGCGATCGTTGCGGGGCTGCGAGATGGCGAACTCGCGTCGGAACAAACCAGTGTCTGTGAGGGACTCTTACGGACCAGGGCATGAGTAAAGAGTCAGGAAATTGTTTTATGAGGTCGACTCAACCCGAAGGTCAGAACTGATGTTATAAAGAAACCATTACAGCGGCTTTGTACACCGAGCCATTAATGGTCGGACGCAGCATTGCAAATCACGTTCTTCTAGCATGTTTGTGGTAAACAGCAG TTTAGCTTAATCATGAGATGTCACGAGTTCCTCGAGTACATCACGGTGGCACTTTAAATAACTCTGCCATTCATAGACATTTTACGTTCCGCAACGTGTTCAATGAGGGCGCGACACATACTCCCCGCGTAACTGGGGGCAGTGATTTGATCTCACCGTACGGACGGTGGAAATCAATGATTTTCAGAACTTCGATCAGTTTCTGGAGGTTCACAGCGTTCTTCTCTGTGTGACTTCCCCTACCCAGATGACATTAGAATTAGTTTGAGCGCAATTTCGACTACCCTTGTGAGGTGCCTCGGCGTTGAAGAACTGGTCGTTGTAGAAA TATAATGATGTATTTAAGCGAAGACAAGAATTTCTGTGGGGCACTTATTTACACTCTGCAATTTGTCGACATGAGACATCCTAGGGGCACGAGGGTGAGTTCCGTGACCAAAGAGCTACATCACAACCCATAGCGACGCGTGTCGAGAATAGAATCAAGACAGCGGCCAGGACTATTAGTACAAATCCTTACAGTATGACCATTCCTCTCATAGCTGTTTTGGCGGGTTTTCGTAGTGCCGGTACCATATCGTGCCTATACAAACTAAGCTGTATCTCCAAAGCCTACTAGCGAACTGCACATTAACTATCCAACACTCCGTACGCAC GCCCGATGCAGCCCTACCGAAACCCAATGTTCAAAGCCTGACTCAGTGAAGAAACTTTCCTCGACAGAAAAACTAGCTCGACATGGGTCTAGTTGCACTTATTATGTGTCGTTTGGGGGGCCGTCCGGGACGGCTTCACCACAGCATTTTCGTTCGGCTAATAGGATCTCCATGAAGTCCTAACGCGCCTACGTTGTCCGCCCTAGTTAGGTATCTCCGGGGTATATAGACGTCGCACTCCAGTGGCCCCGATGGCCGCAGGGTGGTCGTCAGGTTTCCCGCCACACATATATGTTCGCTAATAGACGCACTAAGGTTCACTATGTCA GTGGCCCGTGCATGACTGGACGACTCCTACCATGCGCTGACGCGCAAGGGGGATCACCTGCTGCTTTCCCGATGTCAAGTAATGAAGTACGCACAGAAGTTTTCGAAGGTCCGCTGCCGACTGACTAGAGTGCTGCTTGAGACGACAACTAGGTGTCAGGGGGAAGAATATATCAGCACCTCGCGGGATCCGCCCAGCGAAGGGCGTCCACTTATTATAGGACAAACAGCCTCGGCTCGAAGCCGGCGTATATTCTTGTTGCGCTCCTCCGCATCACTCCAGTGTAACCGCAGGACTGGACACCCTTTGGACGATAGATAGGTAAGCG TTCTTCAGGCGGGCCCGATCGTCCATCGTCACGAGGATGGCTCTCCTAACATACACCAGGTAGTGAAGATTGAGCAGCGCAGTAACATCACTTATTACTCCTCGAGAGAACCTACGTAGTGGCCGCGCGGAAATTCGGAACCGGGTATCACGTCATCGGGGCCGTTCGCCCATTCCAAATTGCAGTGAAGGCCTGTGGAAGTTGCGTCCACACGAAATCTCTGAGAAGTGCATTAATTTGTCCAGACAGGAGAGTAGTAATATAAACAGTAAGACCAACAGCGTATGGCGGACTATCCGTCATAGCAACTGTGCCCTTCGCATATTAT CTCTCCCTGATGGGAATTCACTGGTCACCAAGATTGTCTATACCCCTATGCGGCGCAAGAGCTATTAGTCGGGCACCGTCTACCGCGCGAACGCGTGCGAAGCCGTTATGCACCTCCCCATTTAAGCGGTACCTTTGGTGTACATATTGCACTGGCAAAGATTGTGATCTTTCATGAAGGTATATGGAACAAGGTGCACTAATTCGATGTCCAGAATACACAGTCTGCATGCGGCCAACAATTATAACTCCGCTGGTATGACGCTGTAATATGTTCGAACAGCTTCGCCTGCGTGGAAGAACAATTGTGCACCAGTGCACGCGTTATC TGCACCCGGAATACGTGTTTTTTGCCCCTCGCCGCTTGCACCAATCAGCGCTGGACGAAGAGATATCAGACTATCCTGGGTCCAAATCCGGGTTACTGTAGCCGTTAAATTCCGCTGAATACTGGAGTAACTCCCCTACTATTATAACTCGGTAACCAAGTAATAAGGTAACATGAACTACCAAAGAATGCGCCTTATGGAGGCCAACAGCTAAAAACTATAAGTCTGGCTTGTGATACCGGCGAGGTCCATAGGGTCAGTAATCTGGCCATCTACACCTTGGTCTCAGGTCCGACAAAATCGGTCTAGATTCCCTCATCGAGATTTC GGCGGAGTCTCGCACTGAAGCAATGTCCTTATCCTGACGATACGTATCATCGGCGCTGGAGGTCTCTAATCATATTTCCCAGTGTTTCCGAAAGCCGATAATTCACAACCGCTTTAGGGTCATCTGTTTTGCTCATCATTTTCCTCTCGATAACTTCTCAACATTATCTAGGTTGGAAGTATTACTCTTTGATCCCGCTCGTCTGACAAGTACTTATCGACCACGCGAGGTCCCGTGCCACCATTGTACAAACGGCTATACCTTCTGTTTAGATATTAACTAATGCCAAGAGATACTGTCACTCCATATAACTCGCGTTTGGCCAAGG"
dna_split=dna.split()
kk=15
tt=20
n=2000
BestMotifs=None
BestScore=float('inf')
for i in range(20):
 result=GibbsSample(dna_split,kk,tt,n)
 current_score=Score(result)
 if current_score<BestScore:
     BestScore=current_score
     BestMotifs=result
print(" ".join(BestMotifs))
```
